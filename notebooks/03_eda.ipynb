# Notebook 03 — EDA (Exploratory Data Analysis)

Tất cả biểu đồ đều có **tiêu đề, nhãn trục và nhận xét 2-3 dòng**.
Output lưu tại `reports/figures/`.

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib as mpl
mpl.rcParams['figure.max_open_warning'] = 50
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display, Image

FIG = Path('reports/figures')
FIG.mkdir(parents=True, exist_ok=True)

df = pd.read_csv('data/processed/listings_clean.csv')
print('Loaded cleaned data:', df.shape)
print('Columns sample:', df.columns.tolist()[:10])
```

## 1. Phân bố `price_per_m2` (trước và sau log-transform)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df['price_per_m2'] / 1e6, bins=50, color='steelblue', edgecolor='white')
axes[0].set_title('Phân bố price_per_m2 (thang gốc, đơn vị triệu VND/m²)')
axes[0].set_xlabel('Triệu VND/m²')
axes[0].set_ylabel('Số tin')
axes[1].hist(np.log1p(df['price_per_m2']), bins=50, color='seagreen', edgecolor='white')
axes[1].set_title('Phân bố price_per_m2 (log1p)')
axes[1].set_xlabel('log1p(VND/m²)')
axes[1].set_ylabel('Số tin')
plt.tight_layout()
plt.savefig(FIG / 'fig01_price_distribution.png', dpi=120)
plt.close()
display(Image(filename=str(FIG / 'fig01_price_distribution.png')))
print('Saved fig01')
print('\nNhận xét: Phân bố thang gốc lệch phải (skewed) do vài căn penthouse/căn góc giá > 200tr/m². Log-transform giúp phân bố gần chuẩn — đây là lý do ta fit mô hình trên log1p(price_per_m2) rồi inverse-transform khi đánh giá.')
```

## 2. Top 6 quận có `price_per_m2` trung bình cao nhất

```python
top = (df.groupby('district_clean')['price_per_m2']
         .agg(['mean', 'count']).query('count >= 30')
         .sort_values('mean', ascending=False))
fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(top.index[::-1], top['mean'][::-1] / 1e6, color='coral')
ax.set_title('Giá/m² trung bình theo 6 quận có dữ liệu (≥ 30 tin)')
ax.set_xlabel('Triệu VND/m²')
plt.tight_layout()
plt.savefig(FIG / 'fig02_top_districts.png', dpi=120)
plt.close()
display(Image(filename=str(FIG / 'fig02_top_districts.png')))
print('Saved fig02')
print('\nNhận xét: Quận 7 và Quận Bình Thạnh dẫn đầu — đúng với thực tế BĐS TP.HCM (khu vực phát triển, gần trung tâm, nhiều dự án mới). Quận 12, Bình Tân, Gò Vấp là nhóm giá thấp hơn (vùng ven).')
```

## 3. Boxplot `price_per_m2` theo `direction_clean`

```python
dir_data = df.dropna(subset=['direction_clean'])
order = (dir_data.groupby('direction_clean')['price_per_m2'].median()
         .sort_values(ascending=False).index.tolist())
data = [dir_data.loc[dir_data['direction_clean'] == d, 'price_per_m2'].dropna() / 1e6 for d in order]
fig, ax = plt.subplots(figsize=(10, 5))
ax.boxplot(data, tick_labels=order, showfliers=False)
ax.set_title('Phân bố giá/m² theo hướng nhà (căn hộ)')
ax.set_ylabel('Triệu VND/m²')
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(FIG / 'fig03_price_by_direction.png', dpi=120)
plt.close()
display(Image(filename=str(FIG / 'fig03_price_by_direction.png')))
print('Saved fig03')
print('\nNhận xét: Chênh lệch giá giữa các hướng không lớn (median 46–59 triệu VND/m²). Hướng Đông Bắc và Bắc có median cao nhất (~58–59 triệu/m²); Đông Nam thấp nhất (~46 triệu/m²). Tuy nhiên các hộp overlap nhiều → hướng là feature yếu trong mô hình.')
```

## 4. Scatter `area_m2` vs `total_price`

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(df['area_m2'], df['total_price'] / 1e9, alpha=0.3, s=8, color='purple')
ax.set_title('Diện tích vs tổng giá (căn hộ)')
ax.set_xlabel('Diện tích (m²)')
ax.set_ylabel('Tổng giá (tỷ VND)')
ax.set_xlim(0, 200)
ax.set_ylim(0, 15)
plt.tight_layout()
plt.savefig(FIG / 'fig04_area_vs_price.png', dpi=120)
plt.close()
display(Image(filename=str(FIG / 'fig04_area_vs_price.png')))
print('Saved fig04')
print('\nNhận xét: Quan hệ gần tuyến tính — diện tích càng lớn giá càng cao. Có một số căn diện tích ~50-80m² ở mức giá cao (3-5 tỷ) → căn hộ cao cấp ở vị trí đắt đỏ. Outlier (> 200m² và > 10 tỷ) đã bị lọc trong cleaner.')
```

## 5. Số tin đăng theo quận

```python
counts = df['district_clean'].value_counts()
fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(counts.index, counts.values, color='teal')
ax.set_title('Số tin đăng theo 6 quận')
ax.set_xlabel('Quận')
ax.set_ylabel('Số tin')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig(FIG / 'fig05_listings_by_district.png', dpi=120)
plt.close()
display(Image(filename=str(FIG / 'fig05_listings_by_district.png')))
print('Saved fig05')
print(f'\nNhận xét: Thành phố Thủ Đức có nhiều tin nhất ({counts.iloc[0]} tin, chiếm {counts.iloc[0]/counts.sum()*100:.1f}%) — nhiều dự án mới. Nhóm tiếp theo: Quận 7 ({counts.iloc[1]}), Bình Tân ({counts.iloc[2]}), Bình Thạnh ({counts.iloc[3]}); nhóm ít nhất: Quận 12 ({counts.iloc[4]}), Gò Vấp ({counts.iloc[5]}).')
```

## 6. Heatmap correlation giữa các biến số

```python
num_cols = ['area_m2', 'bedrooms', 'bathrooms', 'direction_code',
            'balcony_code', 'furnishing_code', 'legal_code', 'total_price', 'price_per_m2']
corr = df[num_cols].corr()
fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)
ax.set_xticks(range(len(num_cols)))
ax.set_xticklabels(num_cols, rotation=45, ha='right')
ax.set_yticks(range(len(num_cols)))
ax.set_yticklabels(num_cols)
ax.set_title('Heatmap tương quan Pearson — căn hộ')
for i in range(len(num_cols)):
    for j in range(len(num_cols)):
        ax.text(j, i, f'{corr.iloc[i, j]:.2f}', ha='center', va='center',
                color='white' if abs(corr.iloc[i, j]) > 0.5 else 'black', fontsize=8)
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.savefig(FIG / 'fig06_correlation_heatmap.png', dpi=120)
plt.close()
display(Image(filename=str(FIG / 'fig06_correlation_heatmap.png')))
print('Saved fig06')
print('\nNhận xét: Tương quan giữa price_per_m2 và các biến khác yếu (<0.3) — giá/m² căn hộ phụ thuộc chủ yếu vào vị trí (district) và dự án (project_name) hơn là diện tích/phòng/hướng. Cụm kích thước (area/bedrooms/bathrooms) tương quan mạnh với nhau: bedrooms ↔ area_m2 = 0.74, area_m2 ↔ bathrooms = 0.64, bedrooms ↔ bathrooms = 0.68. total_price ↔ area_m2 = 0.67 — cùng chiều, căn to thì tổng giá cao. direction_code ↔ balcony_code = 0.61 (hướng nào thường đặt ban công hướng đó).')
```

```python
amen = pd.read_csv('data/processed/listings_with_amenities.csv')
amen_raw = pd.read_csv('data/raw/neighborhood_amenities.csv')

# === Panel 1: Bar chart amenity_score trung bình theo quận ===
top_am = (amen.groupby('district_clean')['amenity_score']
           .mean().sort_values(ascending=False))
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

colors = plt.cm.viridis(np.linspace(0.2, 0.85, len(top_am)))
axes[0].bar(top_am.index, top_am.values, color=colors)
axes[0].set_title('Amenity score trung bình theo 6 quận (nguồn OSM thật)', fontsize=12)
axes[0].set_xlabel('Quận')
axes[0].set_ylabel('Amenity score (trung bình)')
axes[0].tick_params(axis='x', rotation=30)
for tick in axes[0].get_xticklabels():
    tick.set_horizontalalignment('right')
# annotate bar values
for i, v in enumerate(top_am.values):
    axes[0].text(i, v + max(top_am.values)*0.01, f'{v:.1f}', ha='center', fontsize=9)

# === Panel 2: Stacked bar — 5 loại tiện ích theo quận (đọc từ raw) ===
amen_cols = ['school_count', 'hospital_count', 'supermarket_count', 'park_count', 'bus_stops_count']
per_district = (amen_raw.drop_duplicates(['district_clean', 'ward'])
                  .groupby('district_clean')[amen_cols].mean()
                  .reindex(top_am.index))
per_district.plot(kind='bar', stacked=True, ax=axes[1],
                   color=['#4C72B0', '#DD8452', '#55A467', '#C44E52', '#8172B2'])
axes[1].set_title('Số lượng tiện ích (mean) theo quận', fontsize=12)
axes[1].set_xlabel('Quận')
axes[1].set_ylabel('Số lượng (count)')
axes[1].tick_params(axis='x', rotation=30)
for tick in axes[1].get_xticklabels():
    tick.set_horizontalalignment('right')
axes[1].legend(title='Loại tiện ích', bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=8)

plt.tight_layout()
plt.savefig(FIG / 'fig09_amenity_by_district.png', dpi=120, bbox_inches='tight')
plt.close()
display(Image(filename=str(FIG / 'fig09_amenity_by_district.png')))
print('Saved fig09 (multi-panel)')
print(f'\nNhận xét: Dữ liệu tiện ích lấy từ OpenStreetMap (Overpass API) thật. Panel trái: TP. Thủ Đức ({top_am.iloc[0]:.1f}) dẫn đầu — gộp 3 quận cũ (Q2, Q9, Thủ Đức) nên tổng tiện ích lớn; kế đến Quận 7 ({top_am.iloc[1]:.1f}) và Quận Bình Tân ({top_am.iloc[2]:.1f}); Quận Gò Vấp ({top_am.iloc[3]:.1f}), Quận 12 ({top_am.iloc[4]:.1f}); thấp nhất Quận Bình Thạnh ({top_am.iloc[5]:.1f}). Panel phải cho thấy cấu trúc khác nhau theo quận: Q7 nổi bật về park_count ({int(per_district.loc["Quận 7", "park_count"])} công viên — gấp 2-3 lần các quận khác nhờ Cầu Ánh Sao, Hồ Bán Nguyệt, Mũi Đèn Đỏ), Bình Tân mạnh về bus_stops + school, Bình Thạnh có ít school nhất do diện tích nhỏ nhất. ⚠️ Lưu ý: TP. Thủ Đức không so sánh được công bằng với 5 quận còn lại vì là gộp 3 quận — nếu tính chia 3, Thủ Đức ≈ {top_am.iloc[0]/3:.1f} thì ngang Quận 7.')
```

## Kết luận EDA

Sau khi phân tích 11 biểu đồ trên tập 1779 tin đăng (có 1773 tin có `price_per_m2` hợp lệ), các phát hiện chính:

**1. Phân bố giá — cần log-transform:**
- `price_per_m2` có **skewness = 4.97** (rất lệch phải), sau khi log-transform giảm xuống **0.76** (gần chuẩn). Do đó mô hình ở Notebook 04 sẽ dùng `log(price_per_m2)` làm target.
- Khoảng giá phổ biến: 30–80 triệu/m². Median = 53 trệu/m².

**2. Vị trí (district) là yếu tố chính:**
- Top 3 quận đắt nhất theo median: **Bình Thạnh (~71 trệu/m²), Quận 7 (~58 trệu/m²), Thủ Đức (~54 trệu/m²)** — chênh lệch ~50–70% so với nhóm cuối (Quận 12, Gò Vấp, Bình Tân ~42–45 trệu/m²).
- Hướng và diện tích có ảnh hưởng nhỏ hơn; **dự án (`project_name`)** cũng quan trọng — có 288 dự án, top là Vinhomes Grand Park (63 tin), The Origami (47), Vinhomes Central Park (44).

**3. Dữ liệu tập trung ngắn hạn (không phân tích xu hướng):**
- Toàn bộ tin đăng trong khoảng **10/07/2026 – 08/08/2026** (~30 ngày): 871 tin tháng 7, 908 tin tháng 8.
- Không đủ độ dài để phân tích xu hướng giá theo thời gian — chỉ dùng để mô tả "ảnh chụp" thị trường.

**4. Cơ cấu phòng ngủ và nội thất:**
- **2PN chiếm 67.4%** (1194/1772) — áp đảo, phù hợp gia đình nhỏ/vợ chồng trẻ. 1PN = 15.3%, 3PN = 16.1%, 4PN = 1.2%.
- Nội thất: 34.3% tin **không mô tả**; trong phần còn lại, "Nội thất cơ bản" dẫn đầu (~32%). Nội thất cao cấp rất hiếm (chỉ 15 tin).

**5. Pháp lý — tín hiệu tích cực:**
- **Sổ hồng riêng chiếm 59.5%** (1059/1779), tiếp theo sổ hồng chung 16.8%. Tỷ lệ "Không rõ" 19.9% — người mua nên xác minh kỹ nhóm này.
